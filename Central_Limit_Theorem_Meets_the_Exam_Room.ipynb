# Central Limit Theorem Meets the Exam Room

```python
# Pakiety
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
from tqdm.notebook import tqdm
```

## The Uniform Distribution

```python
fig, axes = plt.subplots(4, 2, figsize=(12, 10))

n_values = [1, 10, 50, 100]

for i, n in enumerate(n_values):
    obs = np.random.uniform(0, 1, (1000, n))
    means = np.mean(obs, axis=1)

    sns.histplot(obs.ravel(), kde=True, stat="density", color="blue", alpha=0.3, ax=axes[i, 0])
    x = np.linspace(np.min(obs), np.max(obs), 100)
    mu, sigma = np.mean(obs), np.std(obs)
    y = norm.pdf(x, mu, sigma)
    axes[i, 0].plot(x, y, color="black", linestyle="--", linewidth=2, label="Normal PDF")
    axes[i, 0].set_title(f"Histogram and density function (observations), n={n}")
    axes[i, 0].set_xlabel("Values")
    axes[i, 0].set_ylabel("Density")

    sns.histplot(means, kde=True, stat="density", color="green", alpha=0.3, ax=axes[i, 1])
    x = np.linspace(np.min(means), np.max(means), 100)
    mu, sigma = np.mean(means), np.std(means)
    y = norm.pdf(x, mu, sigma)
    axes[i, 1].plot(x, y, color="black", linestyle="--", linewidth=2, label="Normal PDF")
    axes[i, 1].set_title(f"Histogram and density function (means), n={n}")
    axes[i, 1].set_xlabel("Values")
    axes[i, 1].set_ylabel("Density")

plt.tight_layout()
plt.show()
```

## Tests, Luck, and Statistics

### Test-generating function

```python
def test_initiator(n = 10):
    test = np.zeros((n, 3))
    correct_answer_position = []
    for i in range(n):
        random_number =  np.random.choice(range(3))
        test[i, random_number] = 1
        correct_answer_position.append(random_number)
    correct_answer_position = np.array(correct_answer_position)
    return test, correct_answer_position
```

### Function selecting an answer with probability $p$

```python
def random_answers(test, p):
    questions = test[0]
    positions = test[1]
    answers = []
    for i in range(questions.shape[0]):
        question = questions[i,:]
        position = positions[i]
        probs = np.full(3, (1 - p)/2)
        probs[position] = p
        ans = np.random.choice(question, p = list(probs))
        answers.append(ans)
    return np.array(answers)
```

### Experiment

```python
expected_results = np.zeros((100, 100))
stand_err = np.zeros((100, 100))
for i in tqdm(range(1, 101)):
    for j in range(1, 101):
        answers = random_answers(test_initiator(n = i),j/100)
        expected_results[i - 1, j - 1] = np.mean(answers)
        stand_err[i - 1, j - 1] = np.std(answers)/np.sqrt(i)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Results

sns.heatmap(expected_results,
            cmap="coolwarm",
            annot=False, ax=axes[0])
axes[0].set_xticks(np.arange(5, 100, 5))
axes[0].set_xticklabels(np.arange(5, 100, 5)/100)
axes[0].set_yticks(np.arange(5, 100, 5))
axes[0].set_yticklabels(np.arange(5, 100, 5))
axes[0].set_title("Results")

# Variability

sns.heatmap(stand_err,
            cmap="coolwarm",
            annot=False,
            ax=axes[1])
axes[1].set_xticks(np.arange(5, 100, 5))
axes[1].set_xticklabels(np.arange(5, 100, 5)/100)
axes[1].set_yticks(np.arange(5, 100, 5))
axes[1].set_yticklabels(np.arange(5, 100, 5))
axes[1].set_title("Variability")

plt.show()
```

```python
fig, axes = plt.subplots(3, 2, figsize=(12, 10))
x = np.arange(1, 101) / 100


rows = [15, 30, 45, 60, 75, 90]
colors = ['b', 'c', 'g', 'm', 'r', 'y']

for i, row in enumerate(rows):
    ax = axes[i // 2, i % 2]

    y = expected_results[row, :]
    y_min = y - 1.96 * stand_err[row, :]
    y_max = y + 1.96 * stand_err[row, :]

    ax.plot(x, y, label=f"{row}-question test", color=colors[i])
    ax.fill_between(x, y_min, y_max, color=colors[i], alpha=0.2)
    ax.set_title(f"{row}-question test")
    ax.set_xlabel("Probability of correct answer")
    ax.set_ylabel("Test Score")
    ax.set_ylim(0, 1)
    ax.axhline(y=0.6, color='black', linestyle='--')
    ax.legend()

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(3, 2, figsize=(12, 10))
x = np.arange(1, 101)


columns = [40, 45, 50, 55, 60, 65]
colors = ['b', 'c', 'g', 'm', 'r', 'y']

for i, column in enumerate(columns):
    ax = axes[i // 2, i % 2]
    p = column/100
    y = norm.cdf(0.5, loc=column/100, scale=np.sqrt(p * (1 - p))/np.sqrt(np.arange(1,101)))

    ax.plot(x, y, label=f"Probability is equal to {column/100}", color=colors[i])
    ax.set_title(f"The Correct Answer is Selected with Probability {column/100}")
    ax.set_xlabel("Number of questions")
    ax.set_ylabel("The Probability of Failing the Test")
    ax.set_ylim(0, 1)
    ax.axhline(y=0.5, color='black', linestyle='--')
    ax.legend()

plt.tight_layout()
plt.show()
```
